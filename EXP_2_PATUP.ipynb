```python
#2
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#create dataset
data={
    'Age':[22,25,30,35,20,45,39,56],
    'Salary':[30000,25000,40000,45000,50000,68000,35000,42000],
    'Salary_color':['skyblue','lightcoral','lightgreen','gold','plum','orange','red','yellow']
}
df=pd.DataFrame(data)
print(df)
#Histogram
plt.figure(figsize=(10,4))
plt.hist(df['Age'],bins=10, color='blue',edgecolor='green')
plt.title('Age distrtibution')
plt.xlabel('Age')
plt.ylabel('frequency')
plt.show()
#scatter plot
plt.figure(figsize=(6,10))
sns.scatterplot(x='Age', y='Salary', data=df,color='red')
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
#pie chart
plt.figure(figsize=(6,4))
plt.pie(
df['Salary'],
labels=df['Age'],
colors=df['Salary_color'],
autopct='%1.1f%%'
)
plt.title('Salary Distribution by Age')
plt.show()
#Box plot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Salary'],
color='green')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.show()
```
